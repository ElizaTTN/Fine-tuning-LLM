```python
from unsloth import FastLanguageModel
from peft import PeftModel
import torch

# 1. Загружаем базовую модель
model, tokenizer = FastLanguageModel.from_pretrained(
    "unsloth/Phi-3-mini-4k-instruct-bnb-4bit",
    max_seq_length=2048,
    load_in_4bit=True,
)

# 2. Подключаем твой LoRA адаптер (ВАЖНО: путь к папке, не к checkpoint)
model = PeftModel.from_pretrained(
    model,
    "/content/drive/MyDrive/phi3_finetune"
)

# 3. Режим инференса
FastLanguageModel.for_inference(model)
```

```python
# точно рабочий вариант для тестирования по одному запросу

messages = [
    {"role": "user", "content": "Напиши текст в публицистическом стиле. Тема: Глобальное потепление"}
]

inputs = tokenizer.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True,
    return_tensors="pt"
).to("cuda")

outputs = model.generate(
    input_ids=inputs,
    max_new_tokens=300,
    temperature=0.7,
    top_p=0.9,
)

print(tokenizer.decode(outputs[0]))
```

```python
# тестирование по 10 запросов на одну тему
prompts = [
    "Напиши текст в художественном стиле. Тема: Глобальное потепление",
    "Напиши текст в художественном стиле. Тема: Фотосинтез",
    "Напиши текст в художественном стиле. Тема: Вечер перед бурей",
    "Напиши текст в художественном стиле. Тема: Развитие искусственного интеллекта",
    "Напиши текст в художественном стиле. Тема: Заброшенный маяк",
    "Напиши текст в художественном стиле. Тема: Последний поезд ночью",
    "Напиши текст в художественном стиле. Тема: Старое письмо без адреса",
    "Напиши текст в художественном стиле. Тема: Город, который исчез",
    "Напиши текст в художественном стиле. Тема: Одиночество в большом городе",
    "Напиши текст в художественном стиле. Тема: Тайна закрытой двери"

]

for prompt in prompts:
    messages = [{"role": "user", "content": prompt}]

    inputs = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt"
    ).to("cuda")

    outputs = model.generate(
        input_ids=inputs,
        max_new_tokens=300,
        temperature=0.7,
        top_p=0.9,
    )

    print("\n=== PROMPT ===")
    print(prompt)
    print("=== OUTPUT ===")
    print(tokenizer.decode(outputs[0]))
```
